# Churn-Prediction using Neural Network

```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv("C:\\Users\\user\\Datasets\\churn_modelling.csv")
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df=df.drop(columns=['RowNumber','CustomerId','Surname'])
```

```python
df.head()
```

```python
df.duplicated().sum()
```

```python
df['Exited'].value_counts()
```

```python
df['Geography'].value_counts()
```

```python
df['Gender'].value_counts()
```

```python
from sklearn.preprocessing import LabelEncoder
lb=LabelEncoder()
df["Geography"]=lb.fit_transform(df['Geography'])
df["Gender"]=lb.fit_transform(df['Gender'])
```

```python
df.head()
```

```python
x=df.drop(columns=['Exited'],axis=1)
y=df['Exited']
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
x_test.shape
```

```python
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
x_train_sc=sc.fit_transform(x_train)
x_test_sc=sc.fit_transform(x_test)
```

```python
import tensorflow
from tensorflow import keras
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense
```

```python
# ! pip install tensorflow
```

```python
model=Sequential()

model.add(Dense(11,activation='relu',input_dim=10))
model.add(Dense(11,activation='relu'))
model.add(Dense(1,activation='sigmoid'))
```

```python
model.summary()
```

```python
model.compile(loss='binary_crossentropy',optimizer='Adam',metrics=['accuracy'])
```

```python
history=model.fit(x_train_sc,y_train,epochs=100,validation_split=0.2)
```

```python
model.layers[0].get_weights()
```

```python
y_log=model.predict(x_test_sc)
```

```python
y_pred=np.where(y_log>0.5,1,0)
```

```python
from sklearn.metrics import accuracy_score
accuracy=accuracy_score(y_test,y_pred)
```

```python
print(accuracy)
```
